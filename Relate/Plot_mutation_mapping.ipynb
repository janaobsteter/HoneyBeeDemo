```python
uniqueMapC = pd.concat([hitsP1.NoHits.CurrentPol[hitsP1.NoHits.CurrentPol == 1], hitsP1.AF.CurrentPol[hitsP1.NoHits.CurrentPol == 1], hitsP1.NumSamples.CurrentPol[hitsP1.NoHits.CurrentPol == 1]], axis = 1)
uniqueMapC.columns = ["NoHits", "AF", "NumSamples"]
uniqueMapC.loc[:, "Pol"] = "Current"
uniqueMapO = pd.concat([hitsP1.NoHits.OppositePol[hitsP1.NoHits.OppositePol == 1], hitsP1.AF.OppositePol[hitsP1.NoHits.OppositePol == 1], hitsP1.NumSamples.OppositePol[hitsP1.NoHits.OppositePol == 1]], axis = 1)
uniqueMapO.columns = ["NoHits", "AF", "NumSamples"]
uniqueMapO.loc[:, "Pol"] = "Opposite"

```

```python
uniqueMap = pd.concat([uniqueMapC, uniqueMapO])
uniqueMap.head()
uniqueMap.NoHits.value_counts()
```

```python
uniqueMap.NumSamples = [x.split(":")[1].strip("}") for x in uniqueMap.NumSamples]
uniqueMap.head()
```

```python
ggplot() + geom_histogram(aes(x = "AF"), data = uniqueMap) + ggtitle("AF of the mutations that map uniquely")
```

```python
hits.loc[:, "Unique"] = [True if site in uniqueMap.index else False for site in hits.Site ]
hits.loc[:, "Map"] = np.where(((hits.NoHits == 1) & (hits.Unique)), True, False)
hits.head()
```

```python
#Plot the histogram of mutations that map uniquely and all mutations in the mid-frequency
ggplot() + geom_histogram(aes(x = "AF", fill = "Map"),  position = "dodge", data = hits[(hits.AF > 0.2) & (hits.AF < 0.8)], bins = 50) + facet_wrap(facets="Polarisation") + \
    scale_fill_manual(values = ["red", "black"], name = "Map uniquely")
```

```python
# Plot the ratio of uniquely mapping to all mutations
breakpoints = np.linspace(0, 1, 200)

groupsMapped = hits[hits.Map == True].groupby(['Polarisation', pd.cut(hits.AF, breakpoints)])
mappedCount = groupsMapped.size().unstack()

groupsAll = hits.groupby(['Polarisation', pd.cut(hits.AF, breakpoints)])
allCount = groupsAll.size().unstack()
allCount.head()

mappedCountL = mappedCount.melt()
mappedCountL.loc[:, "Polarisation"] = ["Reference", "Alternative"] * int(len(mappedCountL) / 2)
mappedCountL.columns = ["AF", "CountMapped", "Polarisation"]
allCountL = allCount.melt()
allCountL.loc[:, "Polarisation"] = ["Reference", "Alternative"] * int(len(allCountL) / 2)
allCountL.columns = ["AF", "CountAll", "Polarisation"]
ratioMapped = allCountL.merge(mappedCountL, on = ["AF", "Polarisation"])
ratioMapped.loc[:, "Ratio"] = ratioMapped.CountMapped / ratioMapped.CountAll
ratioMapped.head()

ggplot() + geom_point(aes(x = "AF", y = "Ratio"), data = ratioMapped) + facet_wrap(facets="Polarisation") + \
    ggtitle("Ratio of uniquely mapping to all mutations on AF in alternate polarisations")
```
