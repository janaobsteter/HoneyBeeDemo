```python
from lets_plot import *
import pandas as pd
import numpy as np
import os

LetsPlot.setup_html()
```

```python
chromosomes = range(1, 17)
```

```python
for chromosome in chromosomes:
    print(chromosome)
    if os.path.isfile("TracePlot_Diversity" + str(chromosome) + ".csv"):
        tmp = pd.read_csv("TracePlot_Diversity" + str(chromosome) + ".csv")
        if chromosome == 1:
            tracePlots = tmp
        else:
            tracePlots = tracePlots.merge(tmp)
```

```python
tracePlots = tracePlots.rename({'Unnamed: 0': 'Sample'}, axis='columns')
```

```python
tracePlots = tracePlots.melt(id_vars = "Sample", var_name = "Chromosome")
```

```python
tracePlots.head()
```

```python
ggplot(data = tracePlots) + geom_line(aes(x = "Sample", y = "value", color = "Chromosome"))
```
